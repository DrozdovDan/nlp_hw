### Простая разметка и разбиение данных на обучающую и тестовую выборку [2 балла]

Теперь займёмся предобработкой слов.

Для начала будем считать, что выходная последовательность состоит из двух чисел. Цифрой `1` будем указывать границу подстроки, то есть символ, стоящий перед `_`. В остальных случаях заполняем последовательность нулями.

Итак, заполним позиции символов следующим образом:

`0 - <Word-In>`

`1 - <Word-End>`

```python
def simple_compound_annotation(word: str) -> str:
    """
    Данная функция кодирует слово в последовательность из 0 и 1
    1 - <Word-End> - буква, являющаяся границей подслова, т.е. стоящая перед `_`
    0 - <Word-In> - в противном случае
    :param word: слово
    :return строка со списком
    """
    seq = []
    for i in range(len(word) - 1):
        if word[i + 1] == '_':
            seq.append(1)
        elif word[i] == '_':
            continue
        else:
            seq.append(0)
    seq.append(0)
    return str(seq)
```

```python
def list2tensor(l):
    return torch.tensor(l, dtype=torch.long)

def encode_words(words: List[str]) -> List[List[float]]:
    """
    Напишите функцию, кодирующую слова в последовательности символьных кодов
    Обратите внимание на то, чтобы функция умела обрабатывать символы, которых нет в словаре.
    Также не забываёте о том, что все нижние подчёркивания и маркеры для инфиксов должны быть удалены из слов.
    Чтобы быстрее обрабатывать слова, рекомендуется вместо питоновских списков использвать функцию np.zeros(...)
    и затем добавлять код символа по индексу

    :param words: список слов
    :return массив с тензорами, состоящими из символьных кодов
    """
    cleaned = [w.replace("_", "").replace("+", "").lower() for w in words]
    words_X = []
    for w in cleaned:
        words_X.append(np.zeros(len(w)))
        for i, ch in enumerate(w):
            words_X[-1][i] = char2index.get(ch, char2index[None])
    return [list2tensor(word) for word in words_X]

def encode_tags(tags):
    return [list2tensor(eval(ts)) for ts in tags]
```

### Разделитель слов [3 балла]

Построим сегментатор на основе символьной языковой модели.

```python
class BiLSTMTagger(nn.Module):
    def __init__(self, embedding_dim, hidden_dim, vocab_size, tagset_size, dropout=0.2):
        super(BiLSTMTagger, self).__init__()
        self.hidden_dim = hidden_dim

        self.embeddings = nn.Embedding(vocab_size, embedding_dim, padding_idx=0) # строим symbol2vec-матрицу
        self.bilstm = nn.LSTM(embedding_dim, hidden_dim, bidirectional=True, batch_first=True) # двунаправленный LSTM-слой
        self.dense = nn.Linear(2 * hidden_dim, tagset_size) # обычный fc-layer
        self.dropout = nn.Dropout(p=dropout) # позволим нашей модели не переобучаться на трейне :)

    def forward(self, word):
        """
        Напишите код для архитектуры сети
        В конце вам может пригодиться softmax
        """
        embs = self.embeddings(word)
        output, _ = self.bilstm(embs)
        tag_scores = F.softmax(self.dense(self.dropout(output)), dim=-1)
        return tag_scores
```

### Обучение + валидация [1 балл + 1 балл]

```python
EMBEDDING_DIM = 32
HIDDEN_DIM = 32
N_EPOCHS = 3
N_TAGS = 2
```

```python
device = torch.device('cuda:6')
model = BiLSTMTagger(EMBEDDING_DIM, HIDDEN_DIM, len(char2index), N_TAGS).to(device)
loss_function = nn.CrossEntropyLoss()
optimizer = optim.Adam([p for p in model.parameters()])
```

```python
sum(p.numel() for p in model.parameters())
```

```python
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
from tqdm.auto import tqdm, trange

@torch.no_grad()
def print_metrics(model, data, name="", **kw):
    model.eval()
    
    all_preds = []
    all_labels = []
    total_loss = 0.0
    n_batches = 0

    val_x, val_y = encode_words(data['Word'].values), encode_tags(data['Annotation'].values)

    for i, (word_in, targets) in tqdm(enumerate(zip(val_x, val_y)), desc='Validating', total=len(val_x)):
        word_in = word_in.to(device) if device else word_in
        targets = targets.to(device) if device else targets
        tag_scores = model(word_in)
        
        preds = torch.argmax(tag_scores, dim=-1)
        
        all_preds.append(preds.cpu())
        all_labels.append(targets.cpu())
        
        loss = loss_function(tag_scores, targets)
        total_loss += loss.item()
        n_batches += 1

    all_preds = torch.cat(all_preds).numpy()
    all_labels = torch.cat(all_labels).numpy()
    acc = accuracy_score(all_labels, all_preds)
    precision, recall, f1, _ = precision_recall_fscore_support(all_labels, all_preds, average='macro', zero_division=0)
    avg_loss = (total_loss / n_batches)
    header = "Val"
    
    tqdm.write(f"{header} loss: {avg_loss:.4f}")
    tqdm.write(f"{header} accuracy:  {acc:.4f}")
    tqdm.write(f"{header} precision: {precision:.4f}")
    tqdm.write(f"{header} recall:    {recall:.4f}")
    tqdm.write(f"{header} F1-score:  {f1:.4f}")

    return {
        'loss':      avg_loss,
        'accuracy':  acc,
        'precision': precision,
        'recall':    recall,
        'f1':        f1
    }

    # Берите любые метрики, которые хотите оценивать на валидационной выборке
```

```python
print_metrics(model, val, max_len=128)
```

```python
for epoch in trange(N_EPOCHS, desc='Epochs'):
    model.train()
    running_loss = 0.0
    for i, (word_in, targets) in tqdm(enumerate(zip(train_x, train_y)), desc='Training', total=len(train_x)):
        optimizer.zero_grad()
        word_in = word_in.to(device) if device else word_in
        targets = targets.to(device) if device else targets
        tag_scores = model(word_in)
        loss = loss_function(tag_scores, targets)
        loss.backward()
        optimizer.step()

        # print statistics

        running_loss += loss.item()
        if i % 10000 == 9999:
            tqdm.write(f'[{epoch + 1}, {i + 1:5d}/{len(train_x)}] loss: {running_loss / 10000:.3f}')
            running_loss = 0.0

    ### здесь может быть оценена ваша валидационная выборка
    print_metrics(model, val, max_len=128)
```

```python
predicts = [torch.argmax(model(x.to(device)), dim=-1).detach().cpu() for x in test_x]
```

```python
y_true = [l for seq in test_y for l in seq]
y_pred = [l for seq in predicts for l in seq]
```

```python
target_names = ('Word-In', 'Word-End')

print(classification_report(y_true, y_pred, target_names=target_names))
```

### Усложним нашу разметку для инфиксов [1 балл]

Пусть нашему инфиксу будет присваиваться код 2 (`Infix-End`). В итоге мы будем иметь три тега.

```python
def encode_complex_compound_annotation(word):
    """
    Данная функция кодирует слово в последовательность из 0 и 1
    2 - <Infix-End> - буква, являющаяся границей инфикса, т.е. последняя после `+`
    1 - <Word-End> - буква, являющаяся границей подслова, т.е. стоящая перед `_`
    0 - <Word-In> - в противном случае
    :param word: слово
    :return строка со списком
    """
    seq = []
    infix_started = False
    for i in range(len(word) - 1):
        if word[i] == '_':
            continue
        elif word[i] == '+':
            infix_started = True
            continue
        elif word[i + 1] == '_' and not infix_started:
            seq.append(1)
        elif (word[i + 1] == '_' or word[i + 1] == '+') and infix_started:
            infix_started = False
            seq.append(2)
        else:
            seq.append(0)
    if word[-1] != '_' and word[-1] != '+':
        if infix_started:
            seq.append(2)
        else:
            seq.append(0)
    return str(seq)
```

### Придумайте, как можно улучшить качество на втором типе разметки [2 балла]

Идеи

- добавить в архитектуру больше слоёв
- подумать, можно ли здесь применить attention-механизм или crf-слой
- а может здесь имеет смысл дообучение первой модели на новых данных?
- ваши пожелания :)

```python
class TransTagger(nn.Module):
    def __init__(self, embedding_dim, hidden_dim, vocab_size, tagset_size, dropout=0.2):
        super(TransTagger, self).__init__()
        self.hidden_dim = hidden_dim

        self.embeddings = nn.Embedding(vocab_size, embedding_dim, padding_idx=0) # строим symbol2vec-матрицу
        self.attention = nn.MultiheadAttention(embedding_dim, num_heads=4, batch_first=True, kdim=embedding_dim, vdim=embedding_dim)
        self.dense = nn.Linear(embedding_dim, tagset_size) # обычный fc-layer
        self.dropout = nn.Dropout(p=dropout) # позволим нашей модели не переобучаться на трейне :)
        self.norm = nn.LayerNorm(embedding_dim)
        self.ff = nn.Linear(embedding_dim, embedding_dim)

    def forward(self, word):
        """
        Напишите код для архитектуры сети
        В конце вам может пригодиться softmax
        """
        embs = self.embeddings(word)
        output1 = self.norm(self.attention(embs, embs, embs)[0]) + embs
        output2 = self.norm(self.ff(output1) + embs) + output1
        tag_scores = F.softmax(self.dense(self.dropout(output2)), dim=-1)
        return tag_scores
```

```python
device = 'cuda:6'
model = TransTagger(EMBEDDING_DIM, HIDDEN_DIM, len(char2index), N_TAGS).to(device)
loss_function = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

```python
for epoch in trange(N_EPOCHS, desc='Epochs'):
    model.train()
    running_loss = 0.0
    for i, (word_in, targets) in tqdm(enumerate(zip(train_x, train_y)), desc='Training', total=len(train_x)):
        optimizer.zero_grad()
        word_in = word_in.to(device) if device else word_in
        targets = targets.to(device) if device else targets
        tag_scores = model(word_in)
        loss = loss_function(tag_scores, targets)
        loss.backward()
        optimizer.step()

        # print statistics

        running_loss += loss.item()
        if i % 10000 == 9999:
            tqdm.write(f'[{epoch + 1}, {i + 1:5d}/{len(train_x)}] loss: {running_loss / 10000:.3f}')
            running_loss = 0.0

    ### здесь может быть оценена ваша валидационная выборка
    print_metrics(model, full_val, max_len=128)
```
